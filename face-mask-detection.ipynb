```python
import os
```

```python
import cv2
```

```python
from keras.preprocessing import image
```

```python
categories=['with_mask','without_mask']
```

```python
data=[]
for category in categories:
    path=os.path.join('train',category)

    label=categories.index(category)

    for file in os.listdir(path):

        img_path=os.path.join(path,file)
        img=cv2.imread(img_path)
        img=cv2.resize(img,(224,224))

        data.append([img,label])
```

```python
len(data)
```

```python
import random
```

```python
random.shuffle(data)
```

```python
x=[]
y=[]

for features,label in data:
    x.append(features)
    y.append(label)
    
```

```python
len(x)
```

```python
len(y)
```

```python
import numpy as np
```

```python
x=np.array(x)
y=np.array(y)
```

```python
x.shape
```

```python
y.shape
```

```python
x=x/255
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from keras.applications.vgg16 import VGG16
```

```python
vgg=VGG16()
```

```python
vgg.summary()
```

```python
from keras import Sequential
```

```python
model=Sequential()
```

```python
for layer in vgg.layers[:-1]:
    model.add(layer)
```

```python
model.summary()
```

```python
for layer in model.layers:
    layer.trainable=False
```

```python
model.summary()
```

```python
from keras.layers import Dense
```

```python
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,epochs=1,validation_data=(x_test,y_test))
```

```python
def detect_face_mask(img):
    resized = cv2.resize(img, (224, 224))
    # Uncomment the next line if your model was trained on [0, 1] normalized pixels:
    # resized = resized.astype("float32") / 255.0
    inp = resized.reshape(1, 224, 224, 3)
    y_pred = model.predict(inp, verbose=0)
    return y_pred[0][0]
```

```python
def draw_label(img, text, pos, bg_color):
    x, y = pos
    (w, h), baseline = cv2.getTextSize(text, cv2.FONT_HERSHEY_SIMPLEX, 1, 2)

    cv2.rectangle(img, (x, y - h - 4), (x + w + 4, y + baseline), bg_color, cv2.FILLED)
    cv2.putText(img, text, (x + 2, y - 2), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
```
